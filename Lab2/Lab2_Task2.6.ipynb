# TASK 2.6: Mersenne Prime - Số nguyên tố lớn và Các thuật toán số học

**Mục tiêu:** Xây dựng chương trình thực hiện các thao tác số học nền tảng cho hệ thống mật mã khóa công khai:
1. **Sinh số nguyên tố ngẫu nhiên** với kích thước bit tùy chỉnh (8-bit, 16-bit, 64-bit) và kiểm tra tính nguyên tố sử dụng thuật toán kiểm định xác suất **Miller-Rabin**.
2. **Tìm 10 số nguyên tố lớn nhất** nhỏ hơn số nguyên tố Mersenne thứ 10.
3. **Thuật toán Euclid** để tìm Ước số chung lớn nhất (GCD) của hai số nguyên lớn.
4. **Lũy thừa theo module nhanh** ($a^x \pmod p$) xử lý các số mũ rất lớn ($x > 40$).

```python
import random
import math

# 1. Thuật toán Lũy thừa theo module nhanh (Modular Exponentiation)
def modular_pow(base, exponent, modulus):
    """Tính (base^exponent) % modulus hiệu quả với số mũ lớn."""
    if modulus == 1:
        return 0
    result = 1
    base = base % modulus
    while exponent > 0:
        if exponent % 2 == 1:
            result = (result * base) % modulus
        exponent = exponent // 2
        base = (base * base) % modulus
    return result

# 2. Thuật toán kiểm tra số nguyên tố Miller-Rabin
def miller_rabin_test(n, k=5):
    """Kiểm tra tính nguyên tố của n bằng thuật toán Miller-Rabin với k lần kiểm định."""
    if n < 2:
        return False
    if n in (2, 3):
        return True
    if n % 2 == 0:
        return False

    # Viết n - 1 dưới dạng d * 2^r
    r = 0
    d = n - 1
    while d % 2 == 0:
        r += 1
        d //= 2

    # Kiểm định k lần
    for _ in range(k):
        a = random.randint(2, n - 2)
        x = modular_pow(a, d, n)
        if x == 1 or x == n - 1:
            continue
        
        for _ in range(r - 1):
            x = modular_pow(x, 2, n)
            if x == n - 1:
                break
        else:
            return False
    return True

# 3. Hàm sinh số nguyên tố ngẫu nhiên có độ dài b bits
def generate_prime(bits):
    """Sinh một số nguyên tố ngẫu nhiên có độ dài bit chỉ định."""
    while True:
        # Sinh số lẻ ngẫu nhiên có độ dài `bits`
        p = random.getrandbits(bits)
        # Đảm bảo bật bit cao nhất và thấp nhất để đúng số lượng bit và là số lẻ
        p |= (1 << bits - 1) | 1
        if miller_rabin_test(p):
            return p

# 4. Thuật toán Euclid tìm Ước chung lớn nhất (GCD)
def euclidean_gcd(a, b):
    """Tìm ước số chung lớn nhất của hai số nguyên lớn bằng thuật toán Euclid."""
    while b != 0:
        a, b = b, a % b
    return a
```

```python
print("=== 1. SINH SỐ NGUYÊN TỐ NGẪU NHIÊN ===")
p_8bit = generate_prime(8)
p_16bit = generate_prime(16)
p_64bit = generate_prime(64)

print(f"- Số nguyên tố 8-bit  : {p_8bit}")
print(f"- Số nguyên tố 16-bit : {p_16bit}")
print(f"- Số nguyên tố 64-bit : {p_64bit}")

print("\n=== 2. TÌM 10 SỐ NGUYÊN TỐ LỚN NHẤT NHỎ HƠN SỐ MERSENNE THỨ 10 ===")
# Danh sách các số mũ của số nguyên tố Mersenne (M_p = 2^p - 1):
# Các số mũ p tương ứng với 10 số nguyên tố Mersenne đầu tiên là: 
# 2, 3, 5, 7, 13, 17, 19, 31, 61, 89
# Do đó số nguyên tố Mersenne thứ 10 ứng với p = 89, tức là M_89 = 2^89 - 1
m10_mersenne = (1 << 89) - 1
print(f"Số nguyên tố Mersenne thứ 10 (2^89 - 1): {m10_mersenne}")

print("Đang tìm 10 số nguyên tố lớn nhất nhỏ hơn M_10 (quá trình này có thể mất vài giây)...")
found_primes = []
current_num = m10_mersenne - 1 # Bắt đầu từ số chẵn ngay trước đó hoặc trừ dần để tìm số lẻ

# Đảm bảo là số lẻ
if current_num % 2 == 0:
    current_num -= 1

while len(found_primes) < 10:
    if miller_rabin_test(current_num):
        found_primes.append(current_num)
    current_num -= 2

print("\n10 số nguyên tố lớn nhất nhỏ hơn số nguyên tố Mersenne thứ 10 là:")
for idx, prime in enumerate(found_primes, 1):
    print(f"{idx}. {prime}")

print("\n=== 3. KIỂM TRA SỐ NGUYÊN TỐ TÙY Ý NHỎ HƠN 2^89 - 1 ===")
# Thử kiểm tra một số ngẫu nhiên gần m10_mersenne
test_candidate = m10_mersenne - 15
is_prime = miller_rabin_test(test_candidate)
print(f"Số kiểm tra: {test_candidate}")
print(f"Kết quả: {'Là số nguyên tố' if is_prime else 'Không phải số nguyên tố'}")
```

```python
print("=== 4. TÍNH ƯỚC CHUNG LỚN NHẤT (GCD) CỦA HAI SỐ LỚN ===")
# Sử dụng hai số nguyên rất lớn tùy ý
num1 = 12345678901234567890987654321
num2 = 9876543210987654321234567890
gcd_result = euclidean_gcd(num1, num2)

print(f"Số lớn 1: {num1}")
print(f"Số lớn 2: {num2}")
print(f"Ước chung lớn nhất (GCD): {gcd_result}")


print("\n=== 5. TÍNH LŨY THỪA THEO MODULE NHANH (Số mũ lớn x > 40) ===")
# Yêu cầu đề bài: ví dụ 7^40 mod 19
base_val = 7
exp_val = 40
mod_val = 19

result_mod = modular_pow(base_val, exp_val, mod_val)
print(f"Biểu thức: {base_val}^{exp_val} mod {mod_val}")
print(f"Kết quả tính toán: {result_mod}")

# Thử thêm một số mũ cực lớn khác để chứng minh hiệu năng
huge_base = 123456789
huge_exp = 987654321987654321
huge_mod = 1000000007
huge_result = modular_pow(huge_base, huge_exp, huge_mod)
print(f"\nThử nghiệm số mũ cực lớn:")
print(f"({huge_base}^{huge_exp}) mod {huge_mod} = {huge_result}")
```
