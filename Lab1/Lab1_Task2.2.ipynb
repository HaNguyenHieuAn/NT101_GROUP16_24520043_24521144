```python
#TASK 2.2
from collections import Counter
import re

def analyze_frequency(text):
    # Chuẩn hóa: Chuyển thành chữ in hoa và loại bỏ ký tự không phải chữ cái
    cleaned_text = re.sub(r'[^A-Z]', '', text.upper())
    
    # 1. Thống kê Unigram (Từng chữ cái)
    unigrams = Counter(cleaned_text)
    print("--- TẦN SUẤT CHỮ CÁI (UNIGRAMS) ---")
    for letter, count in unigrams.most_common(5):
        print(f"{letter}: {count} lần ({count/len(cleaned_text):.2%})")
        
    # 2. Thống kê Bigram (Cặp 2 chữ cái)
    bigrams = Counter(cleaned_text[i:i+2] for i in range(len(cleaned_text)-1))
    print("\n--- TẦN SUẤT CẶP KÝ TỰ (BIGRAMS) ---")
    for bg, count in bigrams.most_common(5):
        print(f"{bg}: {count} lần")
        
    # 3. Thống kê Trigram (Bộ 3 chữ cái)
    trigrams = Counter(cleaned_text[i:i+3] for i in range(len(cleaned_text)-2))
    print("\n--- TẦN SUẤT BỘ 3 KÝ TỰ (TRIGRAMS) ---")
    for tg, count in trigrams.most_common(5):
        print(f"{tg}: {count} lần")

# Đọc file và chạy phân tích
with open('cipher2.2.txt', 'r', encoding='utf-8') as f:
    ciphertext = f.read()
analyze_frequency(ciphertext)
```

```python
# Bảng khóa ánh xạ đã tìm được qua phân tích tần suất
key_map = {
    'A': 'Q', 'B': 'Y', 'C': 'Z', 'D': 'F', 'E': 'V', 'F': 'U', 'G': 'T', 'H': 'H',
    'I': 'R', 'J': 'X', 'K': 'P', 'L': 'A', 'M': 'N', 'N': 'M', 'O': 'L', 'P': 'K',
    'Q': 'J', 'R': 'I', 'S': 'S', 'T': 'G', 'U': 'W', 'V': 'E', 'W': 'D', 'X': 'C',
    'Y': 'B', 'Z': 'O'
}

def decrypt_monoalphabetic(ciphertext, key_map):
    plaintext = ""
    for char in ciphertext:
        if char.isalpha():
            # Kiểm tra chữ hoa hay thường để giữ đúng định dạng
            if char.isupper():
                plaintext += key_map[char]
            else:
                plaintext += key_map[char.upper()].lower()
        else:
            # Giữ nguyên khoảng trắng, số và dấu câu
            plaintext += char
    return plaintext

# Chạy giải mã
with open('cipher2.2.txt', 'r', encoding='utf-8') as f:
    ciphertext = f.read()

print("--- KẾT QUẢ GIẢI MÃ ---")
print(decrypt_monoalphabetic(ciphertext, key_map))
```
