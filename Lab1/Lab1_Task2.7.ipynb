# Nhiệm vụ 2.7: Mở rộng - Mật mã Affine (Affine Cipher)

### 1. Giới thiệu
Mật mã Affine là một dạng mật mã thay thế đơn bảng (monoalphabetic substitution). Thay vì chỉ dịch chuyển một khoảng cố định (phép cộng) như Caesar cipher, Affine cipher sử dụng một hàm tuyến tính kết hợp cả phép nhân và phép cộng trên tập hợp $Z_{26}$ (bảng chữ cái tiếng Anh).

### 2. Nguyên lý hoạt động
Thuật toán sử dụng một cặp khóa $K = (a, b)$, trong đó:
- $a$: Khóa nhân. Cần thỏa mãn điều kiện $a$ và 26 là hai số nguyên tố cùng nhau, tức là $ƯCLN(a, 26) = 1$. Điều kiện này bắt buộc để đảm bảo tồn tại phần tử nghịch đảo của $a$ trong phép modulo 26, giúp quá trình giải mã có thể thực hiện được.
- $b$: Khóa cộng (độ dịch chuyển), $0 \le b \le 25$.

Quy ước ánh xạ chữ cái $A-Z$ tương ứng với các giá trị số từ $0$ đến $25$.

**Quy tắc Mã hóa (Encryption):**
Hàm mã hóa cho một ký tự bản rõ $P$ được định nghĩa là:
$$C = E(P) = (a \cdot P + b) \pmod{26}$$

**Quy tắc Giải mã (Decryption):**
Hàm giải mã cho một ký tự bản mã $C$ được tính bằng cách nhân hai vế với nghịch đảo modulo của $a$ (ký hiệu là $a^{-1}$):
$$P = D(C) = a^{-1} \cdot (C - b) \pmod{26}$$
Trong đó, $a^{-1}$ là số thỏa mãn: $(a \cdot a^{-1}) \pmod{26} = 1$.

```python
import math

# 1. Hàm tìm nghịch đảo modulo (Multiplicative Inverse)
def mod_inverse(a, m):
    """
    Tìm nghịch đảo modulo của a trong Z_m.
    Trả về a^-1 sao cho (a * a^-1) % m = 1.
    Nếu không tồn tại, trả về None.
    """
    for x in range(1, m):
        if (a * x) % m == 1:
            return x
    return None
```

```python
# 2. Hàm Mã hóa Affine
def affine_encrypt(text, a, b):
    # Kiểm tra điều kiện khóa 'a'
    if math.gcd(a, 26) != 1:
        raise ValueError(f"Khóa a = {a} không hợp lệ. a và 26 phải là nguyên tố cùng nhau.")
    
    result = ""
    for char in text:
        if char.isalpha():
            base = ord('A') if char.isupper() else ord('a')
            p = ord(char) - base
            
            # Công thức: C = (a * P + b) mod 26
            c = (a * p + b) % 26
            result += chr(c + base)
        else:
            result += char 
    return result
```

```python
# 3. Hàm Giải mã Affine
def affine_decrypt(text, a, b):
    # Tìm nghịch đảo của 'a'
    a_inv = mod_inverse(a, 26)
    if a_inv is None:
        raise ValueError(f"Không thể giải mã vì khóa a = {a} không có nghịch đảo modulo 26.")
        
    result = ""
    for char in text:
        if char.isalpha():
            base = ord('A') if char.isupper() else ord('a')
            c = ord(char) - base
            
            # Công thức: P = a^-1 * (C - b) mod 26
            p = (a_inv * (c - b)) % 26
            result += chr(p + base)
        else:
            result += char
    return result
```

```python

if __name__ == "__main__":
    print("=== CHƯƠNG TRÌNH MẬT MÃ AFFINE (AFFINE CIPHER) ===")
    print("1. Mã hóa văn bản")
    print("2. Giải mã văn bản")
    print("3. Chạy ví dụ minh họa tự động (Test case)")
    print("0. Thoát")
    
    while True:
        choice = input("\nNhập lựa chọn của bạn (0-3): ").strip()
        
        if choice == '0':
            print("Đã thoát chương trình.")
            break
            
        elif choice == '1':
            plaintext = input("Nhập bản rõ (Plaintext): ")
            try:
                a = int(input("Nhập khóa nhân 'a' (Gợi ý: 3, 5, 7, 9, 11, 15, 17...): "))
                b = int(input("Nhập khóa cộng 'b' (Từ 0 đến 25): "))
                ciphertext = affine_encrypt(plaintext, a, b)
                print(f"[+] Bản mã (Ciphertext): {ciphertext}")
            except ValueError as e:
                print(f"[-] LỖI: {e}. Vui lòng nhập đúng định dạng số và 'a' phải nguyên tố cùng nhau với 26.")
                
        elif choice == '2':
            ciphertext = input("Nhập bản mã (Ciphertext): ")
            try:
                a = int(input("Nhập khóa nhân 'a' đã dùng để mã hóa: "))
                b = int(input("Nhập khóa cộng 'b' đã dùng để mã hóa: "))
                plaintext = affine_decrypt(ciphertext, a, b)
                print(f"[+] Bản rõ (Plaintext): {plaintext}")
            except ValueError as e:
                print(f"[-] LỖI: {e}")
                
        elif choice == '3':
            # Ví dụ minh họa tĩnh chứng minh tính đúng đắn theo yêu cầu đề bài
            test_text = "INFORMATION SECURITY UIT 2026!"
            test_a = 5
            test_b = 8
            
            print("\n--- CHẠY VÍ DỤ MINH HỌA ---")
            print(f"Bản rõ gốc     : {test_text}")
            print(f"Khóa sử dụng   : a = {test_a}, b = {test_b}")
            
            encrypted = affine_encrypt(test_text, test_a, test_b)
            print(f"Bản mã thu được: {encrypted}")
            
            decrypted = affine_decrypt(encrypted, test_a, test_b)
            print(f"Sau khi giải mã: {decrypted}")
            
            if test_text == decrypted:
                print("=> Chương trình hoạt động bình thường!")
            else:
                print("=> Có lỗi xảy ra trong thuật toán.")
        else:
            print("[-] Lựa chọn không hợp lệ, vui lòng nhập lại!")

```
