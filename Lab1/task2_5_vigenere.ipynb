```python
def mahoa(text, key):
    result = ""
    i = 0
    key = key.upper()  
    for char in text:
        if char.isalpha():
            base = ord('A') if char.isupper() else ord('a')
            
            p = ord(char) - base
            k = ord(key[i % len(key)]) - ord('A')
            
            result += chr((p + k) % 26 + base)
            i += 1  # Chỉ tăng chỉ số khóa khi gặp ký tự chữ cái
        else:
            result += char  # Giữ nguyên dấu cách, số, dấu câu
            
    return result

def giaima(text, key):
    result = ""
    i = 0
    key = key.upper()

    for char in text:
        if char.isalpha():
            base = ord('A') if char.isupper() else ord('a')
            
            c = ord(char) - base
            k = ord(key[i % len(key)]) - ord('A')
            
            # Giải mã: (c - k + 26) % 26
            result += chr((c - k + 26) % 26 + base)
            i += 1
        else:
            result += char
            
    return result

a = int(input("Nhap lua chon (1. Ma hoa | 2. Giai ma): "))

if a == 1:
    m = input("Nhap plaintext: ")
    k = input("Nhap khoa: ")
    c = mahoa(m, k)
    print("Ban ma:", c)
elif a == 2:
    m = input("Nhap ciphertext: ")
    k = input("Nhap khoa: ")
    p = giaima(m, k)
    print("Ban ro:", p)
else:
    print("Lua chon khong hop le!")
```
