# Nhiệm vụ 2.5 - Vigenère
## Quy ước
Biểu diễn $A=0, \dots, Z=25$:
- Mã hóa: $C_i = (p_i + k_{i \bmod m}) \bmod 26$.
- Giải mã: $p_i = (C_i - k_{i \bmod m} + 26) \bmod 26$.
trong đó $m$ là độ dài khóa $k$.

Chỉ mã hóa/giải mã các chữ cái alphabet (`A–Z`, `a–z`), giữ nguyên chữ hoa/thường, khoảng trắng, số và các ký tự đặc biệt.
Khóa $k$ chỉ xét các ký tự alphabet (không phân biệt hoa thường).

```python
def vigenere_clean_key(key):
    if not isinstance(key, str):
        raise ValueError("Phải là chuỗi ký tự.")
    cleaned = [c.upper() for c in key if 'A' <= c <= 'Z' or 'a' <= c <= 'z']
    if not cleaned:
        raise ValueError("Cần nhập ký tự")
    return cleaned


def vigenere_encrypt(text, key):
    if not isinstance(text, str):
        raise ValueError("Phải là chuỗi ký tự.")
    key_letters = vigenere_clean_key(key)
    m = len(key_letters)
    result = []
    key_idx = 0
    for char in text:
        if 'A' <= char <= 'Z':
            shift = ord(key_letters[key_idx % m]) - ord('A')
            result.append(chr((ord(char) - ord('A') + shift) % 26 + ord('A')))
            key_idx += 1
        elif 'a' <= char <= 'z':
            shift = ord(key_letters[key_idx % m]) - ord('A')
            result.append(chr((ord(char) - ord('a') + shift) % 26 + ord('a')))
            key_idx += 1
        else:
            result.append(char)
    return "".join(result)


def vigenere_decrypt(ciphertext, key):
    if not isinstance(ciphertext, str):
        raise ValueError("Phải là chuỗi ký tự")
    key_letters = vigenere_clean_key(key)
    m = len(key_letters)
    result = []
    key_idx = 0
    for char in ciphertext:
        if 'A' <= char <= 'Z':
            shift = ord(key_letters[key_idx % m]) - ord('A')
            result.append(chr((ord(char) - ord('A') - shift) % 26 + ord('A')))
            key_idx += 1
        elif 'a' <= char <= 'z':
            shift = ord(key_letters[key_idx % m]) - ord('A')
            result.append(chr((ord(char) - ord('a') - shift) % 26 + ord('a')))
            key_idx += 1
        else:
            result.append(char)
    return "".join(result)
```

## Kiểm thử Vigenère

```python
import unittest


class TestVigenere(unittest.TestCase):
    def test_known_examples(self):
        examples = [
            ("ATTACKATDAWN", "LEMON", "LXFOPVEFRNHR"),
            ("cryptography", "python", "rprwhbvptwvl"),
            ("Hello World", "KEY", "Rijvs Uyvjn"),
            ("UIT @ 2026! CyberSecurity", "LAB", "FIU @ 2026! NycprTpcvciuj"),
            ("Vigenere Cipher 101", "D", "Yljhqhuh Flskhu 101"),
            ("SECRET DATA", "A", "SECRET DATA"),
            ("SHORT", "VERYLONGCIPHERKEY", "NLFPE"),
            ("XYZxyz", "C", "ZABzab"),
            ("PYTHON", "kEy", "ZCRRSL"),
            ("Meet at 10:30 PM, Room #402. Don't be late!", "PASS",
             "Bewl pt 10:30 HE, Goge #402. Sof'l qe dsie!"),
        ]
        for plaintext, key, ciphertext in examples:
            with self.subTest(plaintext=plaintext, key=key):
                self.assertEqual(vigenere_encrypt(plaintext, key), ciphertext)
                self.assertEqual(vigenere_decrypt(ciphertext, key), plaintext)

    def test_key_normalization(self):
        self.assertEqual(vigenere_clean_key("k E-y! 12 à ß"), list("KEY"))
        self.assertEqual(vigenere_encrypt("Hello", "k E-y!"), "Rijvs")
        self.assertEqual(vigenere_encrypt("Hello", "KEY"), vigenere_encrypt("Hello", "kEy"))

    def test_preservation_and_key_position(self):
        self.assertEqual(vigenere_encrypt("A! à ß Đ 12\nB", "BC"), "B! à ß Đ 12\nD")
        self.assertEqual(vigenere_decrypt("B! à ß Đ 12\nD", "BC"), "A! à ß Đ 12\nB")
        self.assertEqual(vigenere_encrypt("", "KEY"), "")
        self.assertEqual(vigenere_decrypt("", "KEY"), "")
        self.assertEqual(vigenere_encrypt("123!? à", "KEY"), "123!? à")

    def test_round_trip(self):
        samples = ("Quick brown fox jumps over lazy dog.", "Short text", "XXX", "JIGGLE",
                   "ALLCAPS AND lowercase with 12345!? à Đ ß", "A", "", "\nHello\tWorld")
        for key in ("A", "Z", "SECRET", "VERYLONGCIPHERKEYTHATEXCEEDSTEXTLENGTH"):
            for text in samples:
                with self.subTest(text=text, key=key):
                    ciphertext = vigenere_encrypt(text, key)
                    self.assertEqual(vigenere_decrypt(ciphertext, key), text)

    def test_sample_100_words(self):
        self.assertGreaterEqual(len(SAMPLE_PLAINTEXT.split()), 100)
        self.assertEqual(vigenere_decrypt(sample_cipher, SAMPLE_KEY), SAMPLE_PLAINTEXT)

    def test_invalid_inputs(self):
        for transform in (vigenere_encrypt, vigenere_decrypt):
            for key in ("", "12345", "!@#$", "à ß Đ", None, 123, []):
                with self.subTest(transform=transform.__name__, key=key):
                    with self.assertRaises(ValueError):
                        transform("Hello", key)
            for text in (None, 123, []):
                with self.subTest(transform=transform.__name__, text=text):
                    with self.assertRaises(ValueError):
                        transform(text, "KEY")


vigenere_test_result = unittest.TextTestRunner(verbosity=2).run(
    unittest.defaultTestLoader.loadTestsFromTestCase(TestVigenere)
)
if not vigenere_test_result.wasSuccessful():
    raise RuntimeError("Kiểm thử Vigenère thất bại.")
```
